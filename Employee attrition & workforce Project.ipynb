## Dealing With Invalid Datatypes

```python
df.dtypes
```

```python
df.head()
```

```python
df['Monthly_Income'].unique()
```

```python
df['Monthly_Income'] = df['Monthly_Income'].str.strip().str.replace('INR','')
df['Monthly_Income'] = df['Monthly_Income'].str.replace(',','')
```

```python
df['Monthly_Income'] = pd.to_numeric(df['Monthly_Income'], errors='coerce')
```

```python
df['Monthly_Income'] = df['Monthly_Income'].fillna(df['Monthly_Income'].median())
```

```python
df['Monthly_Income'].isna().sum()
```

```python
df['Training_Times_Last_Year'].unique()
```

```python
df['Training_Times_Last_Year'] = df['Training_Times_Last_Year'].fillna(df['Training_Times_Last_Year'].mode())
```

```python
df['Training_Times_Last_Year'] = pd.to_numeric(df['Training_Times_Last_Year'], errors = 'coerce')
```

## Missing Values

```python
df.isnull().sum()
```

```python
(df.isna().sum()/5000)*100 
```

```python
# deleting rows having missing values <= 5
df.dropna(subset=['Marital_Status','Education',
                 'Job_Satisfaction','Work_Life_Balance','Environment_Satisfaction',
                  'Training_Times_Last_Year','Work_Life_Balance','Training_Times_Last_Year'],inplace=True)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Performance_Rating'] = df['Performance_Rating'].fillna(df['Performance_Rating'].median())
```

```python
df['Distance_From_Home_KM'] = df['Distance_From_Home_KM'].fillna(df['Distance_From_Home_KM'].median())
```

```python
df.shape
```

```python
df['Age'] = df['Age'].astype(int)
```

## Cleaning text columns

```python
df['Gender'].unique()
```

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
```

```python
df['Gender'] = df['Gender'].replace({'m':'Male', 'male':'Male', 'f':'Female', 'female':'Female', 'fem':'Female'})
```

```python
df['Department'] = df['Department'].str.strip()
```

```python
df['Department'] = df['Department'].str.title()
```

```python
df['Department'].unique()
```

```python
df['City'] = df['City'].str.strip().str.title()
```

```python
df['Overtime'].unique()
```

```python
df['Overtime'] = df['Overtime'].str.strip().str.upper()
```

```python
df['Overtime'] = df['Overtime'].replace({'YES':'Yes','Y':'Yes','TRUE':'Yes','NO':'No','N':'No','FALSE':'No'})
```

```python
df['Business_Travel'].value_counts()
```

```python
df['Business_Travel'] = df['Business_Travel'].str.title()
```

```python
df['Business_Travel'] = df['Business_Travel'].replace({'Travel_Rarely':'Travel Rarely','Travel_Frequently':'Travel Frequently'})
```

```python
df['Employee_Name'].value_counts()
```

```python
df['Employee_Name'] = df['Employee_Name'].str.strip().str.title()
```

```python
df['Job_Role'].value_counts()
```

```python
df['Attrition'].value_counts()
```

```python
df['Attrition'] = df['Attrition'].str.lower()
```

```python
df['Attrition'] = df['Attrition'].replace({'yes':'Yes','y':'Yes','no':'No','n':'No'})
```
